## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
marks = []
for _ in range(5):
    entry = int(input("Enter marks: "))
    if entry == -1:
        break
    if entry < 0 or entry > 100:
        print("Invalid marks")
        continue
    marks.append(entry)
if marks:
    print("Valid Marks:", marks)
    print("Average =", sum(marks) / len(marks))
```

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
rooms = [101, 102, 103, 104, 105]
booked = []
while len(booked) < len(rooms):
    print("Available rooms:", [r for r in rooms if r not in booked])
    choice = int(input("Choose room: "))
    if choice in booked or choice not in rooms:
        print("Room not available.")
    else:
        booked.append(choice)
        print(f"Room {choice} booked successfully!")
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0.0
while True:
    print("\n--- Banking Menu ---\n1. Create Account\n2. Deposit\n3. Withdraw\n4. Check Balance\n5. Exit")
    choice = input("Select option: ")
    if choice == "1":
        balance = 0.0
        print("Account created!")
    elif choice == "2":
        amt = float(input("Deposit amount: "))
        balance += amt
    elif choice == "3":
        amt = float(input("Withdraw amount: "))
        if amt <= balance:
            balance -= amt
        else:
            print("Insufficient balance")
    elif choice == "4":
        print("Current Balance:", balance)
    elif choice == "5":
        break
```

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for s in range(1, 6):
    marks = [int(input(f"Student {s} Subject {m + 1}: ")) for m in range(4)]
    failed = [m + 1 for m, val in enumerate(marks) if val < 33]
    if failed:
        print(f"Student {s}: Failed in Subject(s) {failed}")
    else:
        avg = sum(marks) / len(marks)
        grade = "A" if avg >= 80 else "B" if avg >= 60 else "C"
        print(f"Student {s}: {marks} → Grade: {grade}")
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
for pkg in range(1, 4):
    ratings = []
    for f in range(1, 4):
        r = float(input(f"Package {pkg} Feature {f} Rating: "))
        if r == 0:
            continue
        ratings.append(r)
    avg = sum(ratings) / len(ratings) if ratings else 0
    print(f"Package {pkg}: Avg = {avg:.2f}")
```
