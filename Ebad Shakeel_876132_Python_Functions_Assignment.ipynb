## Question 12 — Electricity Bill Function

**Problem:**  
An electricity company charges different rates based on units consumed. Create a function that calculates the bill.

**Suggested rates (you may adjust slightly):**
- 0 – 100 units → Rs. 5 per unit
- 101 – 200 units → Rs. 10 per unit
- Above 200 units → Rs. 15 per unit

**Requirements:**
- Create a function that accepts the number of **units consumed**.
- Calculate and **return** the final electricity bill using conditions.
- Display the returned bill outside the function.

**Focus:** Function + arguments + conditions + `return`

````python
def calculate_electricity_bill(units):
    if units < 0:
        raise ValueError("Units consumed cannot be negative.")

    if units <= 100:
        bill = units * 5
    elif units <= 200:
        bill = (100 * 5) + ((units - 100) * 10)
    else:
        bill = (100 * 5) + (100 * 10) + ((units - 200) * 15)

    return bill


units_consumed = 250
bill_amount = calculate_electricity_bill(units_consumed)
print(f"Units consumed : {units_consumed}")
print(f"Electricity bill: Rs. {bill_amount:,.2f}")
````

## Question 16 — Student Marks Processor

**Problem:**  
A teacher has a list of marks of all students in a class and wants key statistics.

**Requirements:**
- Create a function that accepts a **list of student marks**.
- Using a loop (or appropriate logic), calculate and **return**:
  - Highest mark
  - Lowest mark
  - Total marks
  - Average marks
- Display the results outside the function.

**Focus:** Function + list + loop + `return`

````python
def process_student_marks(marks):
    if not marks:
        return None, None, 0, 0  
    highest = marks[0]
    lowest = marks[0]
    total = 0

    for mark in marks:
        total += mark
        if mark > highest:
            highest = mark
        if mark < lowest:
            lowest = mark

    average = total / len(marks)
    return highest, lowest, total, average


class_marks = [78, 92, 65, 88, 74]
highest, lowest, total, average = process_student_marks(class_marks)
print(f"Marks        : {class_marks}")
print(f"Highest mark : {highest}")
print(f"Lowest mark  : {lowest}")
print(f"Total marks  : {total}")
print(f"Average marks: {average:.2f}")
````

## Question 23 — Nested Salary Calculator

**Problem:**  
Salary calculation has several smaller steps. Use nested functions to organize the logic cleanly.

**Requirements:**
- Create an outer function called `calculate_salary()`.
- Inside it, create separate **nested functions** for:
  - Calculating allowance
  - Calculating deduction
  - Calculating final salary
- The outer function should combine the results and return the final salary.
- Call the outer function and display the result.

**Focus:** Nested functions

````python
def calculate_salary(basic_salary, allowance_rate, deduction_rate):
    def calculate_allowance():
        return basic_salary * allowance_rate / 100

    def calculate_deduction():
        return basic_salary * deduction_rate / 100

    def calculate_final_salary():
        return basic_salary + calculate_allowance() - calculate_deduction()

    return calculate_final_salary()


final_salary = calculate_salary(60000, 10, 5)
print(f"Final salary: Rs. {final_salary:,.2f}")
````

## Question 24 — Nested Order Processing

**Problem:**  
An order processing system calculates subtotal, discount, tax, and final amount. Organize this using nested functions.

**Requirements:**
- Create an outer function called `process_order()`.
- Inside it, create nested functions for:
  - Calculate subtotal
  - Calculate discount
  - Calculate tax
  - Calculate final amount
- Return the final order amount from the outer function.
- Demonstrate with sample values.

**Focus:** Nested functions + arguments + `return` + real-world logic

````python
def process_order(products, discount_percentage=10, tax_percentage=5):
    def calculate_subtotal():
        return sum(price * quantity for price, quantity in products)

    def calculate_discount(subtotal):
        return subtotal * discount_percentage / 100

    def calculate_tax(amount_after_discount):
        return amount_after_discount * tax_percentage / 100

    def calculate_final_amount():
        subtotal = calculate_subtotal()
        discount = calculate_discount(subtotal)
        discounted_amount = subtotal - discount
        tax = calculate_tax(discounted_amount)
        return subtotal, discount, tax, discounted_amount + tax

    return calculate_final_amount()


items = [(1500, 2), (800, 1), (500, 3)]
subtotal, discount, tax, final_amount = process_order(items, discount_percentage=10, tax_percentage=5)
print(f"Subtotal : Rs. {subtotal:,.2f}")
print(f"Discount : Rs. {discount:,.2f}")
print(f"Tax      : Rs. {tax:,.2f}")
print(f"Final    : Rs. {final_amount:,.2f}")
````

## Question 27 — Employee Payroll System

**Problem:**  
Build a small employee payroll system using multiple functions.

**Requirements:**
Your program should contain **separate functions** for:
- Employee information
- Basic salary
- Allowance
- Deduction
- Tax
- Final salary calculation

**Rules:**
- Use arguments and `return` properly.
- Use conditions where required (e.g., tax slabs).
- Do **not** write everything inside one function.
- Display a clean final salary report.

**Focus:** Function design + modular programming

````python
def employee_information(name, employee_id, department):
    return {"name": name, "employee_id": employee_id, "department": department}


def calculate_basic_salary(monthly_basic):
    return monthly_basic


def calculate_allowance(basic_salary, allowance_rate=10):
    return basic_salary * allowance_rate / 100


def calculate_deduction(basic_salary, deduction_rate=3):
    return basic_salary * deduction_rate / 100


def calculate_tax(taxable_salary):

    if taxable_salary <= 50000:
        return 0
    elif taxable_salary <= 100000:
        return (taxable_salary - 50000) * 0.05
    else:
        return (50000 * 0.05) + (taxable_salary - 100000) * 0.10


def calculate_final_salary(basic_salary, allowance, deduction, tax):
    return basic_salary + allowance - deduction - tax


employee = employee_information("Ahmed Khan", "SMIT-101", "IT")
basic = calculate_basic_salary(85000)
allowance = calculate_allowance(basic, 12)
deduction = calculate_deduction(basic, 3)
taxable_salary = basic + allowance - deduction
tax = calculate_tax(taxable_salary)
net_salary = calculate_final_salary(basic, allowance, deduction, tax)

print("========== EMPLOYEE PAYROLL REPORT ==========")
for key, value in employee.items():
    print(f"{key.replace('_', ' ').title():<15}: {value}")
print(f"Basic salary    : Rs. {basic:,.2f}")
print(f"Allowance       : Rs. {allowance:,.2f}")
print(f"Deduction       : Rs. {deduction:,.2f}")
print(f"Tax             : Rs. {tax:,.2f}")
print("---------------------------------------------")
print(f"Net salary      : Rs. {net_salary:,.2f}")
````

````python
def get_student_information(**details):
 
    return details


def calculate_total(marks):
    total = 0
    for mark in marks:
        total += mark
    return total


def calculate_percentage(total, number_of_subjects, maximum_per_subject=100):
    maximum_total = number_of_subjects * maximum_per_subject
    return (total / maximum_total) * 100 if maximum_total else 0


def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    else:
        return "F"


def check_pass_fail(marks, pass_mark=40):
   
    passed_all = all((lambda mark: mark >= pass_mark)(mark) for mark in marks)
    return "Pass" if passed_all else "Fail"


def display_result_card(student, subject_marks):
    marks = list(subject_marks.values())
    total = calculate_total(marks)
    percentage = calculate_percentage(total, len(marks))
    grade = calculate_grade(percentage)
    result = check_pass_fail(marks)

    print("============= STUDENT RESULT CARD =============")
    for key, value in student.items():
        print(f"{key.replace('_', ' ').title():<15}: {value}")
    print("-----------------------------------------------")
    for subject, mark in subject_marks.items():
        print(f"{subject:<15}: {mark}/100")
    print("-----------------------------------------------")
    print(f"Total           : {total}/{len(marks) * 100}")
    print(f"Percentage      : {percentage:.2f}%")
    print(f"Grade           : {grade}")
    print(f"Result          : {result}")
    print("================================================")
    return total, percentage, grade, result


student_details = get_student_information(name="Sara Ahmed", roll_no="SMIT-205", course="Python Programming")
subject_marks = {"Python": 88, "Mathematics": 79, "English": 92, "Computer": 85}
result_data = display_result_card(student_details, subject_marks)
````

````python
def get_customer_info(name, city, email="Not provided"):
    return {"name": name, "city": city, "email": email}


def show_products(product_catalog):
    print("Available products:")
    for number, product in enumerate(product_catalog, start=1):
        print(f"{number}. {product['name']} — Rs. {product['price']:,.2f}")


def calculate_line_total(price, quantity):
    return price * quantity


def calculate_subtotal(*line_totals):  
    return sum(line_totals)


def calculate_discount(subtotal, discount_rate=10):
    return subtotal * discount_rate / 100


def calculate_tax(amount, tax_rate=5):
    return amount * tax_rate / 100


def calculate_delivery(subtotal, delivery_fee=250, free_delivery_limit=5000):
    if subtotal >= free_delivery_limit:
        return 0
    return delivery_fee


def calculate_order_amount(subtotal, discount_rate=10, tax_rate=5,
                           delivery_fee=250, free_delivery_limit=5000):

    def get_discount():
        return calculate_discount(subtotal, discount_rate)

    def get_tax(discounted_subtotal):
        return calculate_tax(discounted_subtotal, tax_rate)

    def get_delivery():
        return calculate_delivery(subtotal, delivery_fee, free_delivery_limit)

    discount = get_discount()
    discounted_subtotal = subtotal - discount
    tax = get_tax(discounted_subtotal)
    delivery = get_delivery()
    final_amount = discounted_subtotal + tax + delivery
    return discount, tax, delivery, final_amount


def print_order_summary(customer, ordered_products, discount_rate=10, tax_rate=5):
    show_products(ordered_products)
    line_totals = []

    print("\n================= ORDER SUMMARY =================")
    print(f"Customer : {customer['name']}")
    print(f"City     : {customer['city']}")
    print(f"Email    : {customer['email']}")
    print("--------------------------------------------------")
    print(f"{'Product':<16}{'Price':>10}{'Qty':>7}{'Total':>14}")


    line_total = lambda product: calculate_line_total(product['price'], product['quantity'])

    for product in ordered_products:
        amount = line_total(product)
        line_totals.append(amount)
        print(f"{product['name']:<16}{product['price']:>10,.2f}{product['quantity']:>7}{amount:>14,.2f}")

    subtotal = calculate_subtotal(*line_totals)
    discount, tax, delivery, final_amount = calculate_order_amount(
        subtotal,
        discount_rate=discount_rate,
        tax_rate=tax_rate,
    )

    print("--------------------------------------------------")
    print(f"{'Subtotal':<30} Rs. {subtotal:>12,.2f}")
    print(f"{'Discount (' + str(discount_rate) + '%)':<30} Rs. {discount:>12,.2f}")
    print(f"{'Tax (' + str(tax_rate) + '%)':<30} Rs. {tax:>12,.2f}")
    print(f"{'Delivery':<30} Rs. {delivery:>12,.2f}")
    print("--------------------------------------------------")
    print(f"{'FINAL AMOUNT':<30} Rs. {final_amount:>12,.2f}")
    print("==================================================")
    return final_amount


customer = get_customer_info("Ebad", "Karachi", "ebad@example.com")
product_catalog = [
    {"name": "Keyboard", "price": 2500, "quantity": 1},
    {"name": "Wireless Mouse", "price": 1800, "quantity": 2},
    {"name": "Mouse Pad", "price": 400, "quantity": 1},
]

order_total = print_order_summary(customer, product_catalog)
print(f"\nOrder placed successfully. Payable amount: Rs. {order_total:,.2f}")
````
