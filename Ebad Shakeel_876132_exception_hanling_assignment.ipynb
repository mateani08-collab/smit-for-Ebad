A shopkeeper wants to divide the total daily sales equally among a number of branches.

### Q13. Student Record Registration

A school maintains a list of registered student IDs. A new student cannot be registered with an empty ID or an ID that already exists.

Create a custom exception named `DuplicateStudentError`. Validate the entered ID, raise this custom exception when it already exists, and handle it with a specific message. Also handle an empty ID and show the updated list after a successful registration.

```python
# Q13. Student Record Registration
from unittest.mock import patch

class DuplicateStudentError(Exception):
    """Raised when attempting to register a student ID that already exists."""
    pass

def register_student(registered_ids):
    try:
        student_id = input("Enter new student ID: ").strip()
        if not student_id:
            raise ValueError("Student ID cannot be empty.")
        if student_id in registered_ids:
            raise DuplicateStudentError(f"Student ID '{student_id}' is already registered.")
        registered_ids.append(student_id)
        print(f"Registered '{student_id}'. Updated list: {registered_ids}")
    except DuplicateStudentError as err:
        print(f"Duplicate Error: {err}")
    except ValueError as err:
        print(f"Input Error: {err}")

# Test cases: valid new ID, duplicate ID, blank ID
students = ["S101", "S102", "S103"]
for sid in ["S104", "S102", "   "]:
    with patch("builtins.input", return_value=sid):
        register_student(students)
```

```python
# Q14. Weak Password Rejection
from unittest.mock import patch

class WeakPasswordError(Exception):
    """Raised when a password fails length or digit requirements."""
    pass

def validate_password():
    try:
        password = input("Create a password: ")
        issues = []
        if len(password) < 8:
            issues.append("must be at least 8 characters long")
        if not any(char.isdigit() for char in password):
            issues.append("must contain at least one digit")
        if issues:
            raise WeakPasswordError("Password " + " and ".join(issues) + ".")
        print("Password accepted successfully!")
    except WeakPasswordError as err:
        print(f"Weak Password: {err}")

# Test cases: valid, too short, no digits, both failing
for pwd in ["SecurePass9", "short1", "NoDigitsHere", "abc"]:
    with patch("builtins.input", return_value=pwd):
        validate_password()
```

### Q16. Hotel Room Booking

Create two custom exceptions: `InvalidRoomCountError` and `RoomsUnavailableError`. Validate the requested room count, raise the appropriate exception, handle each exception separately, and reduce the available room count only after a successful booking.

```python
# Q16. Hotel Room Booking
from unittest.mock import patch

class InvalidRoomCountError(Exception):
    """Raised when the requested room count is less than 1."""
    pass

class RoomsUnavailableError(Exception):
    """Raised when requested rooms exceed available rooms."""
    pass

def book_rooms(available_rooms):
    try:
        requested = int(input("Enter number of rooms to book: "))
        if requested < 1:
            raise InvalidRoomCountError("You must book at least 1 room.")
        if requested > available_rooms:
            raise RoomsUnavailableError(f"Only {available_rooms} room(s) available; cannot book {requested}.")
        available_rooms -= requested
        print(f"Booked {requested} room(s). Rooms remaining: {available_rooms}")
    except ValueError:
        print("Input Error: Please enter a whole number for rooms.")
    except InvalidRoomCountError as err:
        print(f"Booking Error: {err}")
    except RoomsUnavailableError as err:
        print(f"Availability Error: {err}")
    return available_rooms

# Test cases: valid, zero rooms, exceeds available, non-integer
rooms = 5
for req in ["2", "0", "4", "two"]:
    with patch("builtins.input", return_value=req):
        rooms = book_rooms(rooms)
```

### Q18. File Reading Safely

A report application asks the user for the name of a text file and tries to read its contents.

Use a `try` block to open and read the requested file. Handle `FileNotFoundError` and `PermissionError` with different messages. Use `finally` to display a message that the file-reading attempt has finished, whether it succeeds or fails. Do not assume that a particular file exists.

```python
# Q18. File Reading Safely
from unittest.mock import patch

def read_report_file():
    filename = input("Enter the name of the text file to read: ").strip()
    try:
        with open(filename, "r", encoding="utf-8") as file:
            content = file.read()
            print(f"File '{filename}' contents:\n{content}")
    except FileNotFoundError:
        print(f"File Error: '{filename}' was not found.")
    except PermissionError:
        print(f"Permission Error: You do not have permission to read '{filename}'.")
    finally:
        print(f"Finished file-reading attempt for '{filename}'.\n")

# Test cases: non-existent file, directory (triggers PermissionError/IsADirectoryError on some OS; let's also mock PermissionError)
with patch("builtins.input", return_value="non_existent_report_123.txt"):
    read_report_file()

with patch("builtins.input", return_value="restricted_secret.txt"), \
     patch("builtins.open", side_effect=PermissionError("Access denied")):
    read_report_file()
```

### Q20. Library Borrowing Rules

A library member may borrow up to three books at a time. A member who already has three borrowed books cannot borrow another one. The requested number of books must be a positive integer.

Create custom exceptions named `BorrowLimitExceededError` and `InvalidBookCountError`. Validate the request, raise the correct exception, handle each case separately, and update the member's borrowed-book count only after a valid request.

```python
# Q20. Library Borrowing Rules
from unittest.mock import patch

class BorrowLimitExceededError(Exception):
    """Raised when borrowing exceeds the maximum limit of 3 books."""
    pass

class InvalidBookCountError(Exception):
    """Raised when the requested book count is not a positive integer."""
    pass

def borrow_books(current_borrowed):
    max_limit = 3
    try:
        raw_count = input("Enter number of books to borrow: ")
        count = int(raw_count)
        if count <= 0:
            raise InvalidBookCountError("Requested book count must be a positive integer.")
        if current_borrowed + count > max_limit:
            raise BorrowLimitExceededError(
                f"Cannot borrow {count} book(s). Currently borrowed: {current_borrowed}/{max_limit}."
            )
        current_borrowed += count
        print(f"Borrowed {count} book(s). Total borrowed now: {current_borrowed}/{max_limit}")
    except ValueError:
        print("Input Error: Please enter a valid whole number.")
    except InvalidBookCountError as err:
        print(f"Invalid Request: {err}")
    except BorrowLimitExceededError as err:
        print(f"Limit Exceeded: {err}")
    return current_borrowed

# Test cases: valid (2), exceeds limit (+2 when already 2), zero/negative, non-integer
borrowed = 1
for req in ["1", "2", "0", "one"]:
    with patch("builtins.input", return_value=req):
        borrowed = borrow_books(borrowed)
```

### Q21. ATM with Multiple Custom Errors

Build a small ATM withdrawal program. The withdrawal must be greater than zero, must not exceed the account balance, and must not exceed a daily withdrawal limit.

Create custom exception classes named `InvalidAmountError`, `InsufficientBalanceError`, and `DailyLimitExceededError`. Ask for a withdrawal amount and validate each rule. Raise the relevant custom exception and handle each one with a different message. Update the balance and daily total only after all validations pass.

```python
# Q21. ATM with Multiple Custom Errors
from unittest.mock import patch

class InvalidAmountError(Exception):
    pass

class InsufficientBalanceError(Exception):
    pass

class DailyLimitExceededError(Exception):
    pass

def atm_withdraw(balance, daily_withdrawn, daily_limit=500.0):
    try:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= 0:
            raise InvalidAmountError("Withdrawal amount must be greater than zero.")
        if amount > balance:
            raise InsufficientBalanceError(f"Insufficient balance (${balance:.2f} available).")
        if daily_withdrawn + amount > daily_limit:
            remaining_limit = daily_limit - daily_withdrawn
            raise DailyLimitExceededError(f"Exceeds daily limit of ${daily_limit:.2f} (${remaining_limit:.2f} left today).")
        balance -= amount
        daily_withdrawn += amount
        print(f"Withdrew ${amount:.2f}. New balance: ${balance:.2f} | Withdrawn today: ${daily_withdrawn:.2f}")
    except ValueError:
        print("Input Error: Please enter a valid numeric amount.")
    except InvalidAmountError as err:
        print(f"Amount Error: {err}")
    except InsufficientBalanceError as err:
        print(f"Balance Error: {err}")
    except DailyLimitExceededError as err:
        print(f"Limit Error: {err}")
    return balance, daily_withdrawn

# Test cases: valid (300), exceeds daily limit (300 more -> 600 > 500), non-positive (-50), exceeds balance (2000)
bal, withdrawn_today = 1200.0, 0.0
for amt in ["300", "300", "-50", "2000"]:
    with patch("builtins.input", return_value=amt):
        bal, withdrawn_today = atm_withdraw(bal, withdrawn_today)
```

```python
# Q22. Bank Transfer Between Two Accounts
class InvalidTransferError(Exception):
    pass

class InsufficientFundsError(Exception):
    pass

def transfer_funds(accounts, sender, receiver, amount):
    try:
        if sender not in accounts or receiver not in accounts:
            raise InvalidTransferError("One or both account IDs do not exist.")
        if sender == receiver:
            raise InvalidTransferError("Sender and receiver accounts must be different.")
        if amount <= 0:
            raise InvalidTransferError("Transfer amount must be greater than zero.")
        if amount > accounts[sender]:
            raise InsufficientFundsError(f"Account '{sender}' has insufficient funds (${accounts[sender]:.2f}).")

        accounts[sender] -= amount
        accounts[receiver] += amount
        print(f"Transferred ${amount:.2f} from {sender} to {receiver}.")
    except (InvalidTransferError, InsufficientFundsError) as err:
        print(f"Transfer Failed: {err}")
    finally:
        print(f"Balances -> {sender}: ${accounts.get(sender, 0):.2f}, {receiver}: ${accounts.get(receiver, 0):.2f}\n")

# Test cases: valid, same account, negative amount, insufficient funds
bank_accounts = {"ACC-1": 800.0, "ACC-2": 300.0}
transfer_funds(bank_accounts, "ACC-1", "ACC-2", 250.0)
transfer_funds(bank_accounts, "ACC-1", "ACC-1", 100.0)
transfer_funds(bank_accounts, "ACC-1", "ACC-2", -50.0)
transfer_funds(bank_accounts, "ACC-1", "ACC-2", 1000.0)
```

### Q23. E-commerce Checkout Validation

An online checkout contains a cart total, a delivery address, and a payment amount. The cart total must be positive, the address cannot be blank, and the payment amount must cover the cart total.

```python
# Q23. E-commerce Checkout Validation
class InvalidCartError(Exception):
    pass

class MissingAddressError(Exception):
    pass

class PaymentAmountError(Exception):
    pass

def process_checkout(cart_total, address, payment_amount):
    try:
        if cart_total <= 0:
            raise InvalidCartError("Cart total must be greater than zero.")
        if not address or not address.strip():
            raise MissingAddressError("Delivery address cannot be blank.")
        if payment_amount < cart_total:
            raise PaymentAmountError(
                f"Payment (${payment_amount:.2f}) is less than cart total (${cart_total:.2f})."
            )
        change = payment_amount - cart_total
        print(f"Order Confirmed! Delivering to '{address.strip()}'. Paid: ${payment_amount:.2f} (Change: ${change:.2f})")
    except InvalidCartError as err:
        print(f"Cart Error: {err}")
    except MissingAddressError as err:
        print(f"Address Error: {err}")
    except PaymentAmountError as err:
        print(f"Payment Error: {err}")

# Test cases: valid, empty cart, blank address, insufficient payment
process_checkout(150.00, "42 Maple Street", 150.00)
process_checkout(0.00, "42 Maple Street", 50.00)
process_checkout(150.00, "   ", 150.00)
process_checkout(150.00, "42 Maple Street", 100.00)
```

Create custom exceptions named `InvalidDistanceError` and `PaymentFailedError`. Handle non-numeric input, raise the relevant custom exceptions for invalid distance or insufficient payment, and display the ride receipt only after a successful payment.

```python
# Q26. Ride-Hailing Fare and Payment
from unittest.mock import patch

class InvalidDistanceError(Exception):
    pass

class PaymentFailedError(Exception):
    pass

def complete_ride():
    base_fare = 3.50
    rate_per_km = 1.80
    try:
        distance = float(input("Enter trip distance (km): "))
        if distance <= 0:
            raise InvalidDistanceError("Trip distance must be greater than zero.")
        total_fare = round(base_fare + (distance * rate_per_km), 2)
        payment = float(input(f"Total fare is ${total_fare:.2f}. Enter payment amount: "))
        if payment < total_fare:
            raise PaymentFailedError(f"Payment of ${payment:.2f} is insufficient for fare ${total_fare:.2f}.")
        change = payment - total_fare
        print(f"Receipt -> Distance: {distance} km | Fare: ${total_fare:.2f} | Paid: ${payment:.2f} | Change: ${change:.2f}")
    except ValueError:
        print("Input Error: Distance and payment must be numeric.")
    except InvalidDistanceError as err:
        print(f"Distance Error: {err}")
    except PaymentFailedError as err:
        print(f"Payment Error: {err}")

# Test cases: valid, zero/negative distance, insufficient payment, non-numeric
for dist, pay in [("10", "25"), ("0", "10"), ("10", "15"), ("five", "20")]:
    with patch("builtins.input", side_effect=[dist, pay]):
        complete_ride()
```

### Q27. Medicine Stock and Expiry Check

A pharmacy sells a medicine only when it is in stock, the requested quantity is positive, and the medicine has not expired. Use sample medicine data in your program.

Create custom exceptions named `MedicineOutOfStockError`, `InvalidMedicineQuantityError`, and `MedicineExpiredError`. Validate the request and expiry date, raise and handle each relevant exception, and reduce stock only when the sale is valid.

```python
# Q27. Medicine Stock and Expiry Check
from datetime import date

class MedicineOutOfStockError(Exception):
    pass

class InvalidMedicineQuantityError(Exception):
    pass

class MedicineExpiredError(Exception):
    pass

def dispense_medicine(inventory, med_name, quantity, today_date):
    try:
        med = inventory.get(med_name)
        if not med or med["stock"] <= 0:
            raise MedicineOutOfStockError(f"'{med_name}' is currently out of stock.")
        if quantity <= 0:
            raise InvalidMedicineQuantityError("Requested medicine quantity must be positive.")
        if med["expiry"] < today_date:
            raise MedicineExpiredError(f"'{med_name}' expired on {med['expiry']} and cannot be sold.")
        if quantity > med["stock"]:
            raise MedicineOutOfStockError(f"Only {med['stock']} units of '{med_name}' available.")

        med["stock"] -= quantity
        print(f"Dispensed {quantity} unit(s) of {med_name}. Remaining stock: {med['stock']}")
    except (MedicineOutOfStockError, InvalidMedicineQuantityError, MedicineExpiredError) as err:
        print(f"Pharmacy Alert: {err}")

# Test cases: valid, expired medicine, out of stock, non-positive quantity
pharmacy_stock = {
    "Paracetamol": {"stock": 30, "expiry": date(2027, 6, 1)},
    "Amoxicillin": {"stock": 10, "expiry": date(2025, 1, 15)},
    "Ibuprofen": {"stock": 0, "expiry": date(2027, 8, 10)},
}
current_day = date(2026, 11, 27)

dispense_medicine(pharmacy_stock, "Paracetamol", 5, current_day)
dispense_medicine(pharmacy_stock, "Amoxicillin", 2, current_day)
dispense_medicine(pharmacy_stock, "Ibuprofen", 1, current_day)
dispense_medicine(pharmacy_stock, "Paracetamol", -2, current_day)
```

### Q28. Bulk Employee Payroll Processing

A company stores several employee records in a list. Each record contains an employee name, hours worked, and hourly rate. One invalid record should not stop the payroll calculation for all other employees.

Create a custom exception named `InvalidPayrollRecordError`. Validate each employee record, raise this exception when required data is missing or values are invalid, and handle the error inside the record-processing loop. Calculate payroll for valid employees and report which records were skipped and why.

```python
# Q28. Bulk Employee Payroll Processing
class InvalidPayrollRecordError(Exception):
    pass

def process_bulk_payroll(records):
    valid_payrolls = []
    skipped_records = []

    for idx, rec in enumerate(records, start=1):
        try:
            name = rec.get("name")
            hours = rec.get("hours")
            rate = rec.get("rate")

            if not name or not str(name).strip():
                raise InvalidPayrollRecordError("Missing employee name.")
            if not isinstance(hours, (int, float)) or hours < 0 or hours > 80:
                raise InvalidPayrollRecordError(f"Invalid hours ({hours}) for {name}.")
            if not isinstance(rate, (int, float)) or rate <= 0:
                raise InvalidPayrollRecordError(f"Invalid hourly rate ({rate}) for {name}.")

            pay = hours * rate
            valid_payrolls.append((name.strip(), pay))
        except InvalidPayrollRecordError as err:
            skipped_records.append((idx, rec, str(err)))

    print("Processed Payroll:")
    for emp_name, gross in valid_payrolls:
        print(f"  - {emp_name}: ${gross:.2f}")

    print("\nSkipped Records:")
    for row_num, raw_rec, reason in skipped_records:
        print(f"  - Row {row_num} ({raw_rec}): {reason}")

# Test cases: mix of valid and invalid employee records
employee_records = [
    {"name": "Alice", "hours": 40, "rate": 30},
    {"name": "", "hours": 35, "rate": 25},
    {"name": "Bob", "hours": -5, "rate": 28},
    {"name": "Clara", "hours": 45, "rate": 32},
    {"name": "David", "hours": 38, "rate": 0},
]
process_bulk_payroll(employee_records)
```

### Q29. Inventory Purchase with Safe Transaction

A warehouse purchase operation receives a product ID and requested quantity. The product must exist, the quantity must be positive, and enough stock must be available. A failed purchase must leave inventory unchanged.

Create custom exceptions named `ProductNotFoundError`, `InvalidQuantityError`, and `InsufficientStockError`. Implement validation before changing stock. Use `try`, `except`, `else`, and `finally` meaningfully: report success only in `else`, handle expected failures in `except`, and display an audit/attempt-finished message in `finally`. Demonstrate that stock remains unchanged after a failed purchase.

```python
# Q29. Inventory Purchase with Safe Transaction
class ProductNotFoundError(Exception):
    pass

class InvalidQuantityError(Exception):
    pass

class InsufficientStockError(Exception):
    pass

def purchase_inventory(inventory, product_id, quantity):
    try:
        if product_id not in inventory:
            raise ProductNotFoundError(f"Product ID '{product_id}' does not exist.")
        if quantity <= 0:
            raise InvalidQuantityError("Requested purchase quantity must be positive.")
        if quantity > inventory[product_id]["stock"]:
            raise InsufficientStockError(
                f"Insufficient stock for '{product_id}' (Requested: {quantity}, Available: {inventory[product_id]['stock']})."
            )
    except (ProductNotFoundError, InvalidQuantityError, InsufficientStockError) as err:
        print(f"Transaction Rolled Back: {err}")
    else:
        inventory[product_id]["stock"] -= quantity
        print(f"Purchase Succeeded: {quantity} x {inventory[product_id]['name']} purchased.")
    finally:
        print(f"Audit Log: Purchase attempt finished for '{product_id}'. Current inventory: {inventory}\n")

# Test cases: valid purchase, product not found, insufficient stock, invalid quantity
warehouse = {
    "P01": {"name": "Laptop Stand", "stock": 12},
    "P02": {"name": "Mechanical Keyboard", "stock": 5},
}
purchase_inventory(warehouse, "P01", 3)
purchase_inventory(warehouse, "P99", 1)
purchase_inventory(warehouse, "P02", 10)
purchase_inventory(warehouse, "P02", 0)
```

### Q30. Mini Banking System with a Custom Exception Hierarchy

Design a menu-based mini banking program that supports deposit, withdrawal, and balance viewing. Deposits and withdrawals must be positive. A withdrawal cannot exceed the available balance. Invalid menu choices must also be handled.

Create a base custom exception named `BankingError`, then create `InvalidAmountError` and `InsufficientBalanceError` as child classes of `BankingError`. Raise the appropriate exception from the relevant operation and handle the base class or specific child classes safely. Keep the program running after an invalid transaction, and show the final account balance when the user chooses to exit.

```python
# Q30. Mini Banking System with a Custom Exception Hierarchy
from unittest.mock import patch

class BankingError(Exception):
    """Base exception for banking operations."""
    pass

class InvalidAmountError(BankingError):
    """Raised when deposit or withdrawal amount is non-positive."""
    pass

class InsufficientBalanceError(BankingError):
    """Raised when withdrawal exceeds available balance."""
    pass

def run_mini_bank():
    balance = 1000.00
    while True:
        print("\n--- Mini Bank Menu: [1] Deposit  [2] Withdraw  [3] View Balance  [4] Exit ---")
        choice = input("Select an option (1-4): ").strip()

        if choice == "4":
            print(f"Exiting Mini Bank. Final Account Balance: ${balance:.2f}")
            break
        elif choice == "3":
            print(f"Current Balance: ${balance:.2f}")
        elif choice in ("1", "2"):
            try:
                amount = float(input("Enter amount: "))
                if amount <= 0:
                    raise InvalidAmountError("Amount must be greater than zero.")
                if choice == "1":
                    balance += amount
                    print(f"Deposited ${amount:.2f}. New Balance: ${balance:.2f}")
                else:
                    if amount > balance:
                        raise InsufficientBalanceError(
                            f"Cannot withdraw ${amount:.2f}; available balance is ${balance:.2f}."
                        )
                    balance -= amount
                    print(f"Withdrew ${amount:.2f}. New Balance: ${balance:.2f}")
            except ValueError:
                print("Input Error: Please enter a valid numeric amount.")
            except BankingError as err:
                print(f"Banking Error: {err}")
        else:
            print("Invalid menu choice. Please select 1, 2, 3, or 4.")

# Simulated interactive session testing valid deposit, invalid menu, negative amount, overdraft, balance check, and exit
simulated_inputs = ["1", "250", "9", "2", "-40", "2", "5000", "2", "300", "3", "4"]
with patch("builtins.input", side_effect=simulated_inputs):
    run_mini_bank()
```
